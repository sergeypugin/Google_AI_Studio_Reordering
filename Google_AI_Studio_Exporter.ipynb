```python
#@title Exporter Google AI Studio chat into Obsidian Markdown
#@markdown Настройте параметры и нажмите кнопку **Play** слева для начала работы:

include_thoughts = False #@param {type:"boolean"}

import os
import json
import sys

repo_name = "Google_AI_Studio_Reordering"
if not os.path.exists(repo_name):
    print("Подготовка компонентов...")
    !git clone https://github.com/sergeypugin/Google_AI_Studio_Reordering.git > /dev/null 2>&1

sys.path.append(f'/content/{repo_name}/src')

from google.colab import files
from chat_converter import ChatParser

print("\nВыберите один или несколько JSON-файлов чатов из Google AI Studio:")
uploaded = files.upload()

if not uploaded:
    print("Файлы не были выбраны.")
else:
    for filename, content in uploaded.items():
        try:
            json_data = json.loads(content.decode('utf-8'))
            is_valid, msg = ChatParser.is_a_chat(json_data)
            if not is_valid:
                print(f"Файл `{filename}` не является чатом AI Studio: {msg}")
                continue
            success, md_content = ChatParser.convert_chat_to_md(
                json_data=json_data,
                chat_title=filename,
                file_id="colab_file",
                thoughts_needed=include_thoughts
            )
            if success:
                out_filename = f"colab_{filename}.md"
                with open(out_filename, 'w', encoding='utf-8') as f:
                    f.write(md_content)

                print(f"Успешно! Скачиваем `{out_filename}`...")
                files.download(out_filename)
            else:
                print(f"Ошибка конвертации файла `{filename}`: {md_content}")

        except Exception as e:
            print(f"Ошибка обработки `{filename}`: {e}")
```
